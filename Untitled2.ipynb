```python
# ============================================================================
# CELL 4: Create Biometric Model with Triplet Loss
# ============================================================================

import torch.nn as nn
import torch.nn.functional as F
from torchvision import models

class BiometricEmbeddingNetwork(nn.Module):
    """
    Network that creates unique biometric embeddings
    Similar to face recognition systems (FaceNet, ArcFace)
    """
    def __init__(self, embedding_dim=256):
        super().__init__()

        # Use ResNet50 as backbone (better features than ResNet18)
        resnet = models.resnet50(pretrained=True)

        # Remove final FC layer
        self.backbone = nn.Sequential(*list(resnet.children())[:-1])

        # Embedding layer - compresses 2048 features to embedding_dim
        self.embedding = nn.Sequential(
            nn.Linear(2048, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, embedding_dim),
            nn.BatchNorm1d(embedding_dim)
        )

    def forward(self, x):
        # Extract features
        features = self.backbone(x)
        features = features.view(features.size(0), -1)

        # Get embedding
        embedding = self.embedding(features)

        # L2 normalize (crucial for metric learning)
        embedding = F.normalize(embedding, p=2, dim=1)

        return embedding

class TripletLoss(nn.Module):
    """
    Triplet loss for metric learning
    Makes embeddings of same dog close, different dogs far apart
    """
    def __init__(self, margin=0.5):
        super().__init__()
        self.margin = margin

    def forward(self, anchor, positive, negative):
        # Euclidean distance
        pos_dist = F.pairwise_distance(anchor, positive)
        neg_dist = F.pairwise_distance(anchor, negative)

        # Triplet loss: want pos_dist < neg_dist by at least margin
        loss = F.relu(pos_dist - neg_dist + self.margin)

        return loss.mean()

# Create model
model = BiometricEmbeddingNetwork(embedding_dim=256)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

print("✅ Biometric model created!")
print(f"   Architecture: ResNet50 → 2048 → 512 → 256 (embedding)")
print(f"   Device: {device}")
```

```python
# ============================================================================
# CELL 5: Prepare Dataset with Triplet Sampling
# ============================================================================

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import random
import numpy as np

class TripletDataset(Dataset):
    """
    Dataset that creates triplets: (anchor, positive, negative)
    - Anchor: reference image
    - Positive: same dog, different image
    - Negative: different dog
    """
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform

        # Build index
        self.dog_to_images = {}
        self.dog_ids = []

        for dog_id in sorted(os.listdir(root_dir)):
            dog_path = os.path.join(root_dir, dog_id)
            if os.path.isdir(dog_path):
                images = [os.path.join(dog_path, f) for f in os.listdir(dog_path)
                         if f.endswith(('.jpg', '.jpeg', '.png'))]
                if len(images) >= 2:  # Need at least 2 images for positive pair
                    self.dog_to_images[dog_id] = images
                    self.dog_ids.append(dog_id)

        print(f"📊 Triplet Dataset:")
        print(f"   Dogs: {len(self.dog_ids)}")
        print(f"   Total images: {sum(len(imgs) for imgs in self.dog_to_images.values())}")

    def __len__(self):
        # Each epoch goes through all images once
        return sum(len(imgs) for imgs in self.dog_to_images.values())

    def __getitem__(self, idx):
        # Randomly select anchor dog
        anchor_dog = random.choice(self.dog_ids)

        # Get anchor and positive (same dog, different images)
        if len(self.dog_to_images[anchor_dog]) >= 2:
            anchor_img, positive_img = random.sample(self.dog_to_images[anchor_dog], 2)
        else:
            anchor_img = positive_img = random.choice(self.dog_to_images[anchor_dog])

        # Get negative (different dog)
        negative_dog = random.choice([d for d in self.dog_ids if d != anchor_dog])
        negative_img = random.choice(self.dog_to_images[negative_dog])

        # Load images
        anchor = Image.open(anchor_img).convert('RGB')
        positive = Image.open(positive_img).convert('RGB')
        negative = Image.open(negative_img).convert('RGB')

        # Apply transforms
        if self.transform:
            anchor = self.transform(anchor)
            positive = self.transform(positive)
            negative = self.transform(negative)

        return anchor, positive, negative

# Data augmentation for training
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# No augmentation for testing
test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Create dataset
triplet_dataset = TripletDataset(DATASET_PATH, transform=train_transform)

# Create dataloader
train_loader = DataLoader(
    triplet_dataset,
    batch_size=16,  # Small batch for Colab GPU
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

print(f"✅ DataLoader ready!")
print(f"   Batch size: 16")
print(f"   Batches per epoch: {len(train_loader)}")
```

```python
# ============================================================================
# CELL 6: Train with Triplet Loss
# ============================================================================

import torch.optim as optim
from tqdm import tqdm

# Training setup
EPOCHS = 50  # More epochs for better embeddings
LEARNING_RATE = 0.0001

triplet_loss = TripletLoss(margin=0.5)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)

print("="*70)
print("🚀 STARTING TRAINING - METRIC LEARNING")
print("="*70)
print(f"\nEpochs: {EPOCHS}")
print(f"Learning Rate: {LEARNING_RATE}")
print(f"Margin: 0.5")
print(f"Expected time: ~2-3 hours\n")

train_losses = []
best_loss = float('inf')

for epoch in range(EPOCHS):
    model.train()
    epoch_loss = 0.0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}")
    for anchor, positive, negative in pbar:
        anchor = anchor.to(device)
        positive = positive.to(device)
        negative = negative.to(device)

        # Forward pass
        anchor_emb = model(anchor)
        positive_emb = model(positive)
        negative_emb = model(negative)

        # Calculate triplet loss
        loss = triplet_loss(anchor_emb, positive_emb, negative_emb)

        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()
        pbar.set_postfix({'loss': f'{loss.item():.4f}'})

    # Calculate average loss
    avg_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_loss)

    scheduler.step()

    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # Save best model
    if avg_loss < best_loss:
        best_loss = avg_loss
        torch.save(model.state_dict(), '/content/best_biometric_model.pth')
        print(f"   ✅ Saved best model (loss={avg_loss:.4f})")

print("\n🎉 Training complete!")
print(f"Best loss: {best_loss:.4f}")

# Plot training curve
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(train_losses, marker='o')
plt.xlabel('Epoch')
plt.ylabel('Triplet Loss')
plt.title('Training Progress - Metric Learning')
plt.grid(True)
plt.savefig('/content/training_curve.png', dpi=150, bbox_inches='tight')
plt.show()

print("📊 Training curve saved!")
```

```python
# ============================================================================
# CELL 7: Extract Biometric Embeddings for All Dogs
# ============================================================================

print("\n🔍 Extracting biometric embeddings for all dogs...")

# Load best model
model.load_state_dict(torch.load('/content/best_biometric_model.pth'))
model.eval()

# Extract embeddings
dog_embeddings = {}
dog_to_label = {}

all_dogs = [d for d in os.listdir(DATASET_PATH)
           if os.path.isdir(os.path.join(DATASET_PATH, d))]

with torch.no_grad():
    for idx, dog_id in enumerate(tqdm(all_dogs, desc="Extracting")):
        dog_path = os.path.join(DATASET_PATH, dog_id)
        embeddings_list = []

        # Get all images for this dog
        images = [f for f in os.listdir(dog_path)
                 if f.endswith(('.jpg', '.jpeg', '.png'))]

        for img_name in images:
            img_path = os.path.join(dog_path, img_name)

            try:
                img = Image.open(img_path).convert('RGB')
                img_tensor = test_transform(img).unsqueeze(0).to(device)

                # Extract embedding
                embedding = model(img_tensor)
                embeddings_list.append(embedding.cpu().numpy())
            except Exception as e:
                print(f"⚠️ Error with {img_path}: {e}")
                continue

        if embeddings_list:
            # Average embeddings from multiple images
            avg_embedding = np.mean(embeddings_list, axis=0)
            dog_embeddings[dog_id] = avg_embedding
            dog_to_label[dog_id] = idx

print(f"\n✅ Extracted embeddings for {len(dog_embeddings)} dogs")

# Save embeddings
np.save('/content/biometric_embeddings.npy', dog_embeddings)
print("📁 Saved: /content/biometric_embeddings.npy")
```

```python
# ============================================================================
# CELL 8: Build FAISS Index for Fast Similarity Search
# ============================================================================

print("\n🔧 Building FAISS index for fast retrieval...")

try:
    import faiss

    # Prepare embeddings matrix
    embedding_list = []
    id_list = []

    for dog_id in sorted(dog_embeddings.keys()):
        embedding_list.append(dog_embeddings[dog_id].flatten())
        id_list.append(dog_id)

    embeddings_matrix = np.vstack(embedding_list).astype('float32')

    # Create FAISS index (using cosine similarity)
    # Normalize for cosine similarity
    faiss.normalize_L2(embeddings_matrix)

    # Build index
    dimension = embeddings_matrix.shape[1]
    index = faiss.IndexFlatIP(dimension)  # Inner product (cosine after normalization)
    index.add(embeddings_matrix)

    # Save index
    faiss.write_index(index, '/content/biometric_index.faiss')

    print(f"✅ FAISS index built!")
    print(f"   Dimension: {dimension}")
    print(f"   Total dogs: {len(id_list)}")
    print(f"   Index type: Cosine similarity")

    FAISS_AVAILABLE = True

except ImportError:
    print("⚠️ FAISS not available, will use numpy similarity")
    FAISS_AVAILABLE = False
    index = None
    id_list = sorted(dog_embeddings.keys())
```

```python
# ============================================================================
# CELL 10: NEW DOG REGISTRATION FUNCTION
# ============================================================================

def register_new_dog(image_paths, dog_info, save_to_drive=True):
    """
    Register a new dog with multiple nose images

    Args:
        image_paths: List of 3-5 image paths of the dog's nose
        dog_info: Dictionary with dog information:
            {
                'name': str,
                'breed': str,
                'age': int,
                'gender': str (Male/Female),
                'color': str,
                'weight_kg': int,
                'owner_name': str,
                'owner_contact': str,
                'owner_email': str,
                'owner_address': str,
                'microchip_id': str (optional),
                'medical_notes': str (optional),
                'behavior_notes': str (optional)
            }
        save_to_drive: Save updated database to Google Drive

    Returns:
        dict: Registration result with dog_id and confidence metrics
    """

    print("="*70)
    print("🐕 REGISTERING NEW DOG")
    print("="*70)

    # Validation
    if len(image_paths) < 3:
        return {
            'success': False,
            'error': 'Minimum 3 images required for registration',
            'message': 'Please provide at least 3 clear nose images from different angles'
        }

    if len(image_paths) > 10:
        return {
            'success': False,
            'error': 'Too many images',
            'message': 'Maximum 10 images allowed. Please select 3-5 best quality images'
        }

    # Required fields check
    required_fields = ['name', 'breed', 'age', 'gender', 'owner_name', 'owner_contact']
    missing = [f for f in required_fields if f not in dog_info or not dog_info[f]]

    if missing:
        return {
            'success': False,
            'error': 'Missing required information',
            'missing_fields': missing,
            'message': f'Please provide: {", ".join(missing)}'
        }

    model.eval()

    print(f"\n📸 Processing {len(image_paths)} images...")

    # Extract embeddings from all images
    embeddings_list = []
    valid_images = 0

    for idx, img_path in enumerate(image_paths, 1):
        try:
            # Load image
            img = Image.open(img_path).convert('RGB')
            img_tensor = test_transform(img).unsqueeze(0).to(device)

            # Extract embedding
            with torch.no_grad():
                embedding = model(img_tensor).cpu().numpy()
                embeddings_list.append(embedding)

            valid_images += 1
            print(f"   ✅ Image {idx}/{len(image_paths)} processed")

        except Exception as e:
            print(f"   ⚠️  Image {idx} failed: {e}")
            continue

    if valid_images < 3:
        return {
            'success': False,
            'error': 'Insufficient valid images',
            'message': f'Only {valid_images} images processed successfully. Need at least 3.'
        }

    # Calculate average embedding (robust representation)
    avg_embedding = np.mean(embeddings_list, axis=0)

    # Calculate embedding consistency (quality metric)
    embedding_std = np.std(embeddings_list, axis=0).mean()
    consistency_score = 1 / (1 + embedding_std)  # Higher = more consistent

    print(f"\n📊 Embedding Quality:")
    print(f"   Valid images: {valid_images}")
    print(f"   Consistency score: {consistency_score:.3f}")

    if consistency_score < 0.3:
        print("   ⚠️  Warning: Low consistency - images may be too different")
        print("   💡 Tip: Use images from same session with similar lighting")

    # Generate unique dog ID
    existing_ids = list(dog_embeddings.keys())
    dog_id = f"DOG{len(existing_ids) + 1:04d}"

    # Check for duplicates (important!)
    print(f"\n🔍 Checking for duplicates...")
    duplicate_check = identify_dog_biometric_internal(avg_embedding, top_k=1, threshold=0.85)

    if duplicate_check and duplicate_check[0]['confidence'] > 85:
        match = duplicate_check[0]
        print(f"\n⚠️  WARNING: Possible duplicate detected!")
        print(f"   Existing dog: {match['dog_id']} - {match['profile']['name']}")
        print(f"   Similarity: {match['confidence']:.2f}%")
        print(f"   Owner: {match['profile']['owner_name']}")

        return {
            'success': False,
            'error': 'Duplicate dog detected',
            'duplicate_id': match['dog_id'],
            'duplicate_name': match['profile']['name'],
            'similarity': match['confidence'],
            'message': 'This dog may already be registered. Please verify.'
        }

    # Add embedding to database
    dog_embeddings[dog_id] = avg_embedding

    # Create profile with defaults
    profile = {
        'dog_id': dog_id,
        'name': dog_info['name'],
        'breed': dog_info['breed'],
        'age': dog_info['age'],
        'gender': dog_info['gender'],
        'color': dog_info.get('color', 'Unknown'),
        'weight_kg': dog_info.get('weight_kg', 0),
        'owner_name': dog_info['owner_name'],
        'owner_contact': dog_info['owner_contact'],
        'owner_email': dog_info.get('owner_email', ''),
        'owner_address': dog_info.get('owner_address', ''),
        'microchip_id': dog_info.get('microchip_id', f"LK{random.randint(100000000, 999999999)}"),
        'registration_date': datetime.now().strftime('%Y-%m-%d'),
        'last_seen': datetime.now().strftime('%Y-%m-%d'),
        'vaccinations': dog_info.get('vaccinations', []),
        'medical_notes': dog_info.get('medical_notes', 'No medical notes'),
        'behavior_notes': dog_info.get('behavior_notes', 'No behavior notes'),
        'num_registration_images': valid_images,
        'embedding_consistency': float(consistency_score)
    }

    profiles[dog_id] = profile

    # Update FAISS index
    if FAISS_AVAILABLE and index is not None:
        # Add to index
        new_emb = avg_embedding.flatten().reshape(1, -1).astype('float32')
        faiss.normalize_L2(new_emb)
        index.add(new_emb)
        id_list.append(dog_id)

        # Save updated index
        faiss.write_index(index, '/content/biometric_index.faiss')

    # Save updated databases
    np.save('/content/biometric_embeddings.npy', dog_embeddings)

    with open('/content/dog_profiles.json', 'w') as f:
        json.dump(profiles, f, indent=2)

    # Save to Google Drive
    if save_to_drive:
        output_folder = "/content/drive/MyDrive/dog_biometric_system_FINAL"
        os.makedirs(output_folder, exist_ok=True)

        shutil.copy('/content/biometric_embeddings.npy',
                   f'{output_folder}/embeddings.npy')
        shutil.copy('/content/dog_profiles.json',
                   f'{output_folder}/profiles.json')

        if FAISS_AVAILABLE:
            shutil.copy('/content/biometric_index.faiss',
                       f'{output_folder}/faiss_index.faiss')

        print(f"\n💾 Saved to Google Drive")

    print("\n" + "="*70)
    print("✅ REGISTRATION SUCCESSFUL!")
    print("="*70)
    print(f"🆔 Dog ID: {dog_id}")
    print(f"📛 Name: {profile['name']}")
    print(f"🐕 Breed: {profile['breed']}")
    print(f"👤 Owner: {profile['owner_name']}")
    print(f"📞 Contact: {profile['owner_contact']}")
    print(f"📊 Images used: {valid_images}")
    print(f"✨ Quality score: {consistency_score:.3f}")
    print(f"🗄️  Total dogs in database: {len(dog_embeddings)}")
    print("="*70)

    return {
        'success': True,
        'dog_id': dog_id,
        'profile': profile,
        'images_processed': valid_images,
        'consistency_score': consistency_score,
        'total_dogs_in_db': len(dog_embeddings)
    }

print("✅ Registration function loaded!")
```

```python
# ============================================================================
# CELL 11: IMPROVED IDENTIFICATION FUNCTIONS
# ============================================================================

def identify_dog_biometric_internal(query_embedding, top_k=5, threshold=0.7):
    """
    Internal identification function using embedding
    """
    if FAISS_AVAILABLE and index is not None:
        query_norm = query_embedding.reshape(1, -1).astype('float32')
        faiss.normalize_L2(query_norm)

        distances, indices = index.search(query_norm, top_k)

        results = []
        for dist, idx in zip(distances[0], indices[0]):
            if idx < len(id_list):
                dog_id = id_list[idx]
                similarity = float(dist)
                confidence = similarity * 100

                if similarity >= threshold:
                    results.append({
                        'dog_id': dog_id,
                        'confidence': confidence,
                        'similarity': similarity,
                        'profile': profiles[dog_id]
                    })
    else:
        query_flat = query_embedding.flatten()
        similarities = {}

        for dog_id, stored_emb in dog_embeddings.items():
            stored_emb_flat = stored_emb.flatten()

            similarity = np.dot(query_flat, stored_emb_flat) / (
                np.linalg.norm(query_flat) * np.linalg.norm(stored_emb_flat)
            )

            similarities[dog_id] = float(similarity)

        sorted_matches = sorted(similarities.items(), key=lambda x: x[1], reverse=True)

        results = []
        for dog_id, similarity in sorted_matches[:top_k]:
            confidence = similarity * 100

            if similarity >= threshold:
                results.append({
                    'dog_id': dog_id,
                    'confidence': confidence,
                    'similarity': similarity,
                    'profile': profiles[dog_id]
                })

    return results


def identify_dog_improved(image_path, top_k=5, threshold=0.70, show_below_threshold=True):
    """
    Improved identification with clear "NOT FOUND" handling

    Args:
        image_path: Path to dog nose image
        top_k: Number of top matches to return
        threshold: Minimum confidence for positive identification (0-1)
        show_below_threshold: Show top matches even if below threshold

    Returns:
        dict: Identification result
    """

    print("\n" + "="*70)
    print("🔍 DOG IDENTIFICATION IN PROGRESS...")
    print("="*70)

    model.eval()

    try:
        # Load and preprocess image
        img = Image.open(image_path).convert('RGB')
        img_tensor = test_transform(img).unsqueeze(0).to(device)

        # Extract embedding
        with torch.no_grad():
            query_embedding = model(img_tensor).cpu().numpy().flatten()

        print("✅ Image processed successfully")

    except Exception as e:
        print(f"❌ Error processing image: {e}")
        return {
            'success': False,
            'found': False,
            'error': str(e),
            'message': 'Failed to process image. Please check image quality.'
        }

    # Get matches
    results = identify_dog_biometric_internal(
        query_embedding,
        top_k=top_k,
        threshold=threshold
    )

    # Also get top match regardless of threshold (for display)
    all_results = identify_dog_biometric_internal(
        query_embedding,
        top_k=top_k,
        threshold=0.0
    )

    best_confidence = all_results[0]['confidence'] if all_results else 0

    # Determine if dog is found
    if not results or best_confidence < (threshold * 100):
        print("\n" + "="*70)
        print("❌ DOG NOT FOUND IN DATABASE")
        print("="*70)
        print(f"\n⚠️  No confident match found (threshold: {threshold*100:.1f}%)")

        if all_results:
            print(f"\n📊 Best match confidence: {best_confidence:.2f}%")

            if show_below_threshold and best_confidence > 50:
                print(f"\n💡 Closest matches (below threshold):")
                for i, r in enumerate(all_results[:3], 1):
                    print(f"   {i}. {r['profile']['name']} (ID: {r['dog_id']}) - {r['confidence']:.2f}%")
                print(f"\n   These are NOT confident matches.")

        print(f"\n📝 This dog is likely not registered.")
        print(f"💡 Would you like to register this dog?")
        print("="*70)

        # Show query image
        plt.figure(figsize=(10, 6))
        plt.imshow(img)
        plt.title("❌ DOG NOT FOUND IN DATABASE\n"
                 f"Best Match: {best_confidence:.1f}% (Below {threshold*100:.0f}% threshold)",
                 fontsize=14, fontweight='bold', color='red', pad=20)
        plt.axis('off')
        plt.tight_layout()
        plt.show()

        return {
            'success': True,
            'found': False,
            'best_confidence': best_confidence,
            'threshold': threshold * 100,
            'message': 'Dog not found in database',
            'top_uncertain_matches': all_results[:3] if show_below_threshold else []
        }

    # Dog found!
    print("\n" + "="*70)
    print("✅ DOG IDENTIFIED!")
    print("="*70)

    best_match = results[0]
    p = best_match['profile']

    print(f"\n🐕 IDENTIFICATION RESULT:")
    print(f"   Name: {p['name']}")
    print(f"   ID: {best_match['dog_id']}")
    print(f"   Confidence: {best_match['confidence']:.2f}%")
    print(f"   Breed: {p['breed']}")

    print(f"\n👤 OWNER:")
    print(f"   Name: {p['owner_name']}")
    print(f"   Contact: {p['owner_contact']}")
    print(f"   Email: {p['owner_email']}")

    print(f"\n📋 DETAILS:")
    print(f"   Age: {p['age']} years")
    print(f"   Gender: {p['gender']}")
    print(f"   Microchip: {p['microchip_id']}")
    print(f"   Registered: {p['registration_date']}")

    if len(results) > 1:
        print(f"\n📊 Other potential matches:")
        for r in results[1:4]:
            print(f"   - {r['profile']['name']} (ID: {r['dog_id']}) - {r['confidence']:.2f}%")

    print("="*70)

    # Visualize result
    plt.figure(figsize=(12, 8))
    plt.imshow(img)

    title_color = 'darkgreen' if best_match['confidence'] > 80 else 'orange'
    plt.title(f"✅ IDENTIFIED: {p['name']} (ID: {best_match['dog_id']})\n"
              f"Confidence: {best_match['confidence']:.2f}% | Breed: {p['breed']}\n"
              f"Owner: {p['owner_name']} | Contact: {p['owner_contact']}",
              fontsize=14, fontweight='bold', color=title_color, pad=20)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    return {
        'success': True,
        'found': True,
        'dog_id': best_match['dog_id'],
        'confidence': best_match['confidence'],
        'profile': p,
        'all_matches': results
    }

print("✅ Identification functions loaded!")
```

```python
# ============================================================================
# CELL 13: COMPREHENSIVE SYSTEM TESTING
# ============================================================================

def test_system_comprehensive(num_tests=100, threshold=0.70):
    """
    Comprehensive testing on multiple dogs with proper metrics

    Tests:
    1. True Positive: Correctly identifies dogs in database
    2. False Negative: Fails to identify dog in database
    3. Rank-1 and Rank-5 accuracy
    """

    print("\n" + "="*70)
    print("🧪 COMPREHENSIVE SYSTEM TESTING")
    print("="*70)

    test_sample = min(num_tests, len(dog_embeddings))
    test_dogs = random.sample(list(dog_embeddings.keys()), test_sample)

    # Metrics
    true_positives = 0  # Correctly identified
    false_negatives = 0  # In DB but not found
    rank1_correct = 0
    rank5_correct = 0
    confidences = []

    print(f"\nTesting {test_sample} dogs...")
    print(f"Confidence threshold: {threshold*100:.0f}%\n")

    for dog_id in tqdm(test_dogs, desc="Testing"):
        dog_path = os.path.join(DATASET_PATH, dog_id)

        if not os.path.exists(dog_path):
            continue

        test_images = [f for f in os.listdir(dog_path)
                      if f.endswith(('.jpg', '.jpeg', '.png'))]

        if not test_images:
            continue

        # Use a random image
        test_img = random.choice(test_images)
        test_img_path = os.path.join(dog_path, test_img)

        try:
            # Load and process
            img = Image.open(test_img_path).convert('RGB')
            img_tensor = test_transform(img).unsqueeze(0).to(device)

            with torch.no_grad():
                query_embedding = model(img_tensor).cpu().numpy().flatten()

            # Get results
            results = identify_dog_biometric_internal(query_embedding, top_k=5, threshold=threshold)
            all_results = identify_dog_biometric_internal(query_embedding, top_k=5, threshold=0.0)

            if results:
                confidences.append(results[0]['confidence'])

                # True positive if correct ID
                if results[0]['dog_id'] == dog_id:
                    true_positives += 1
                    rank1_correct += 1

                # Check if in top 5
                top5_ids = [m['dog_id'] for m in results[:5]]
                if dog_id in top5_ids:
                    rank5_correct += 1
            else:
                # Check all results (below threshold)
                if all_results:
                    top5_ids = [m['dog_id'] for m in all_results[:5]]
                    if dog_id in top5_ids:
                        rank5_correct += 1

                # False negative - dog exists but not found
                false_negatives += 1

        except Exception as e:
            print(f"Error testing {dog_id}: {e}")
            continue

    # Calculate metrics
    total_tested = true_positives + false_negatives

    if total_tested > 0:
        accuracy = (true_positives / total_tested) * 100
        fnr = (false_negatives / total_tested) * 100
        rank1_acc = (rank1_correct / total_tested) * 100
        rank5_acc = (rank5_correct / total_tested) * 100
        avg_confidence = np.mean(confidences) if confidences else 0

        print("\n" + "="*70)
        print("📊 TEST RESULTS")
        print("="*70)
        print(f"\n✅ True Positives (Correct):     {true_positives}/{total_tested}")
        print(f"❌ False Negatives (Missed):      {false_negatives}/{total_tested}")
        print(f"\n📈 Metrics:")
        print(f"   Identification Accuracy:  {accuracy:.2f}%")
        print(f"   False Negative Rate:      {fnr:.2f}%")
        print(f"   Rank-1 Accuracy:          {rank1_acc:.2f}%")
        print(f"   Rank-5 Accuracy:          {rank5_acc:.2f}%")
        print(f"   Average Confidence:       {avg_confidence:.2f}%")
        print(f"   Confidence Threshold:     {threshold*100:.0f}%")
        print("="*70)

        # Save metrics
        metrics = {
            'accuracy': accuracy,
            'false_negative_rate': fnr,
            'rank1_accuracy': rank1_acc,
            'rank5_accuracy': rank5_acc,
            'avg_confidence': avg_confidence,
            'true_positives': true_positives,
            'false_negatives': false_negatives,
            'total_tested': total_tested,
            'threshold': threshold * 100
        }

        with open('/content/test_metrics.json', 'w') as f:
            json.dump(metrics, f, indent=2)

        print("\n💾 Metrics saved to: /content/test_metrics.json")

        return metrics
    else:
        print("❌ No valid tests completed")
        return None
```
